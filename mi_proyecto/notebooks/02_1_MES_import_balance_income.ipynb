```python
from pathlib import Path
import shutil

# Asumiendo que ya tienes definidas estas rutas
# RAW_DIR = .../raw/chicago_fed/xport_zip
# INTERIM_DIR = .../interim/chicago_fed/xport

for xpt_path in RAW_DIR.glob("*.xpt"):
    dest = INTERIM_DIR / xpt_path.name
    shutil.copy2(xpt_path, dest)  # o move si prefieres
    print("Copiado a interim:", dest)
```

```python
import pandas as pd

XPORT_DIR = INTERIM_DIR  # o RAW_DIR si no copiaste nada

for xpt_path in XPORT_DIR.glob("*.xpt"):
    print("Leyendo", xpt_path.name)
    df = pd.read_sas(xpt_path, format="xport", encoding="latin1")  # lectura XPORT[web:212][web:216]

    # Limpieza básica opcional
    df.columns = df.columns.str.lower()

    base = xpt_path.stem.lower()
    out_parquet = PROCESSED_DIR / f"{base}.parquet"
    out_csv = PROCESSED_DIR / f"{base}.csv"

    df.to_parquet(out_parquet, index=False)
    df.to_csv(out_csv, index=False)

    print("Guardado:", out_parquet)
```
