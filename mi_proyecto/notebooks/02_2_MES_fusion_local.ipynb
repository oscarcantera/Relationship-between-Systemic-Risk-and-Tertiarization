```python
import pandas as pd
from pathlib import Path

# Paths
path_bal = Path(r"D:\tfg_finance\mi_proyecto_datos\local_db\processed\chicago_fed\call-reports-balance-sheets-1996on.parquet")
path_inc = Path(r"D:\tfg_finance\mi_proyecto_datos\local_db\processed\chicago_fed\call-reports-income-statements-1996on.parquet")

# Cargar balances
df_bal = pd.read_parquet(path_bal)

# Cargar solo claves + columna necesaria del income
cols_inc = ["id_rssd", "date", "ytdnet_operating_earn"]
df_inc = pd.read_parquet(path_inc, columns=cols_inc)
```

```python
# Asegurar tipos
df_bal["date"] = pd.to_datetime(df_bal["date"])
df_inc["date"] = pd.to_datetime(df_inc["date"])

df_bal["id_rssd"] = df_bal["id_rssd"].astype("int32")
df_inc["id_rssd"] = df_inc["id_rssd"].astype("int32")

# Orden (no es obligatorio, pero ayuda a revisar)
df_bal = df_bal.sort_values(["id_rssd", "date"])
df_inc = df_inc.sort_values(["id_rssd", "date"])
```

```python
path_merged = path_bal.with_name("call-reports-balances-plus-income-1996on.parquet")

df_merged = df_bal.merge(
    df_inc,
    on=["id_rssd", "date"],
    how="left",
    validate="one_to_one",   # avisa si hay duplicados en las claves
)

df_merged.info(memory_usage="deep")
df_merged.to_parquet(path_merged, index=False)
print(path_merged)
```

```python
path = Path(r"D:\tfg_finance\mi_proyecto_datos\local_db\processed\chicago_fed\call-reports-balances-plus-income-q1996on.parquet")
panel = pd.read_parquet(path)

# Por seguridad, aseguramos tipos básicos
panel["date"] = pd.to_datetime(panel["date"])
panel["id_rssd"] = panel["id_rssd"].astype("int32")

# Para evitar problemas de división por 0, usamos .where(...)
eps = 1e-9
assets_safe = panel["assets"].where(panel["assets"] != 0, eps)
ln_tot_safe = panel["ln_tot_gross"].where(panel["ln_tot_gross"] != 0, eps)
```

```python
# CAR: equity / assets
panel["CAR"] = panel["equity"] / assets_safe

# LIQ: (cash + securities + trad_ass + ffrepo_ass + ffsold + repo_purch) / assets
panel["LIQ"] = (
    panel["cash"]
    + panel["securities"]
    + panel["trad_ass"]
    + panel["ffrepo_ass"]
    + panel["ffsold"]
    + panel["repo_purch"]
) / assets_safe

# NPL: npl_tot / ln_tot_gross
panel["NPL"] = panel["npl_tot"] / ln_tot_safe

# CIL: ln_ci / assets
panel["CIL"] = panel["ln_ci"] / assets_safe

# HOL: (res_loans_3mo_less + res_loans_3y_5y + res_loans_5y_15y + res_loans_over_15y) / assets
panel["HOL"] = (
    panel["res_loans_3mo_less"]
    + panel["res_loans_3y_5y"]
    + panel["res_loans_5y_15y"]
    + panel["res_loans_over_15y"]
) / assets_safe

# WFUND: (ffpurch + repo_sold + othbor_liab + subdebt) / assets
panel["WFUND"] = (
    panel["ffpurch"]
    + panel["repo_sold"]
    + panel["othbor_liab"]
    + panel["subdebt"]
) / assets_safe

# ROA: ytdnet_operating_earn / assets
panel["ROA"] = panel["ytdnet_operating_earn"] / assets_safe
```

```python
panel.info(memory_usage="deep")

out_path = path.with_name("Variables_MES_banco-trimestre.parquet")
panel.to_parquet(out_path, index=False)
print(out_path)
```
